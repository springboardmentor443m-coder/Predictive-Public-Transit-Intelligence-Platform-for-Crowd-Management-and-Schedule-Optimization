# Namma Metro Station Hourly Ridership — Dataset Analysis

Initial exploratory analysis of the station-hourly ridership dataset.

```python
%pip install pandas

import pandas as pd
from pathlib import Path

dataset_relative_path = Path("datasets/raw/station-hourly.csv")
search_locations = [Path.cwd(), *Path.cwd().parents]

DATASET_PATH = next(
	(
		candidate
		for location in search_locations
		for candidate in [location / dataset_relative_path]
		if candidate.is_file()
	),
	None,
)

if DATASET_PATH is None:
	DATASET_PATH = next(
		(
			candidate
			for location in search_locations
			for candidate in location.rglob("station-hourly.csv")
			if candidate.is_file()
		),
		None,
	)

if DATASET_PATH is None:
	raise FileNotFoundError(
		"Could not find station-hourly.csv in the project directories."
	)

df = pd.read_csv(DATASET_PATH, sep=";")

print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
print("Columns:", list(df.columns))
```

```python
# Preview the data
df.head(10)
```

```python
# Data types and missing values
print("Data types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Duplicate records
print(f"Duplicate rows: {df.duplicated().sum():,}")
```

```python
# Date, hour and station information
print(f"Date range: {df['Date'].min()} to {df['Date'].max()}")
print(f"Number of stations: {df['Station'].nunique():,}")
print(f"Hour range: {df['Hour'].min()} to {df['Hour'].max()}")
```

```python
# Ridership statistics
df["Ridership"].describe()
```

```python
# Highest-ridership station-hour records
df.nlargest(10, "Ridership")[["Date", "Hour", "Station", "Ridership"]]
```

```python
# Average ridership by hour
hourly_avg = (
    df.groupby("Hour")["Ridership"]
      .mean()
      .sort_values(ascending=False)
)
hourly_avg
```

```python
# Average ridership by station
station_avg = (
    df.groupby("Station")["Ridership"]
      .mean()
      .sort_values(ascending=False)
)
station_avg.head(20)
```
