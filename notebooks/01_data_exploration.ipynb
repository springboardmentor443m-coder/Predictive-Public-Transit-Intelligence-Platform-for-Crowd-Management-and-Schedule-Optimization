# MetroFlow — Step 1 Exploratory Data Analysis
Taipei MRT hourly OD dataset. Uses reusable logic from `src/metroflow/data/`.

```python
import sys
sys.path.append('../src')
import pandas as pd
import matplotlib.pyplot as plt
from metroflow.data.loader import load_raw_dataset, discover_raw_files
from config.settings import COL_DATE, COL_HOUR, COL_ENTRY_STATION, COL_EXIT_STATION, COL_TRIP_COUNT

print(f'{len(discover_raw_files())} monthly files available')
```

```python
# Load a small sample of months for interactive exploration (full dataset is ~640M rows).
df = load_raw_dataset(file_limit=3)
df.head()
```

```python
hourly_totals = df.groupby(COL_HOUR)[COL_TRIP_COUNT].sum()
hourly_totals.plot(kind='bar', figsize=(10, 4), title='Total trips by hour of day (sample months)')
plt.xlabel('Hour')
plt.ylabel('Total passenger trips')
plt.show()
```

```python
station_inflow = df.groupby(COL_EXIT_STATION)[COL_TRIP_COUNT].sum().sort_values(ascending=False)
station_inflow.head(15).plot(kind='barh', figsize=(8, 6), title='Top 15 stations by total exit (inflow) volume')
plt.gca().invert_yaxis()
plt.show()
```
