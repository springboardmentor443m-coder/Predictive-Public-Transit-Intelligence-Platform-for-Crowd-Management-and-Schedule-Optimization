```python
print("Number of trips by stop:")

stop_trip_counts = stop_times["stop_id"].value_counts()

display(stop_trip_counts.head(10))
```

```python
top_stops = stop_trip_counts.head(10).reset_index()
top_stops.columns = ["stop_id", "trip_count"]

top_stops["stop_id"] = top_stops["stop_id"].astype(str)
stops["stop_id"] = stops["stop_id"].astype(str)

top_stops = top_stops.merge(
    stops[["stop_id", "stop_name"]],
    on="stop_id",
    how="left"
)

display(top_stops)
```

```python
print("Stop sequence statistics:")

print(stop_times["stop_sequence"].describe())
```

```python
trip_stop_counts = stop_times.groupby("trip_id")["stop_id"].count()

print("Stops per trip:")
print(trip_stop_counts.describe())
```

```python
# Calculate trip duration

trip_times = stop_times.groupby("trip_id").agg(
    start_time=("departure_time", "min"),
    end_time=("arrival_time", "max")
)

print("Sample trip times:")
display(trip_times.head())

print("\nNumber of trips:", len(trip_times))
```

```python
# Calculate trip duration in minutes

trip_times["duration_minutes"] = (
    pd.to_timedelta(trip_times["end_time"].astype(str))
    - pd.to_timedelta(trip_times["start_time"].astype(str))
).dt.total_seconds() / 60

print("Trip duration statistics:")
print(trip_times["duration_minutes"].describe())
```

```python
# Calculate average trip duration by route

trip_route = trips[["trip_id", "route_id"]].drop_duplicates()

route_duration = trip_route.merge(
    trip_times[["duration_minutes"]],
    left_on="trip_id",
    right_index=True,
    how="left"
)

route_duration_summary = (
    route_duration.groupby("route_id")["duration_minutes"]
    .agg(["count", "mean", "min", "max"])
    .sort_values("mean", ascending=False)
)

print("Average trip duration by route:")
display(route_duration_summary)
```

```python
# Add route names to the duration summary

route_duration_named = route_duration_summary.reset_index().merge(
    routes[["route_id", "route_short_name", "route_long_name"]],
    on="route_id",
    how="left"
)

route_duration_named = route_duration_named[
    ["route_id", "route_short_name", "route_long_name",
     "count", "mean", "min", "max"]
]

print("Trip duration by route:")
display(route_duration_named)
```

```python
# Top 5 longest routes by average duration

top_5_longest = route_duration_named.sort_values(
    "mean", ascending=False
).head(5)

print("Top 5 longest routes:")
display(top_5_longest)
```

```python
# Calculate travel time between consecutive stops

stop_times_sorted = stop_times.sort_values(
    ["trip_id", "stop_sequence"]
).copy()

stop_times_sorted["arrival"] = pd.to_timedelta(
    stop_times_sorted["arrival_time"].astype(str)
)

stop_times_sorted["prev_arrival"] = (
    stop_times_sorted.groupby("trip_id")["arrival"].shift(1)
)

stop_times_sorted["stop_to_stop_minutes"] = (
    stop_times_sorted["arrival"] - stop_times_sorted["prev_arrival"]
).dt.total_seconds() / 60

print("Stop-to-stop travel time statistics:")
print(stop_times_sorted["stop_to_stop_minutes"].describe())
```

```python
# Find unusually long gaps between consecutive stops

long_gaps = stop_times_sorted[
    stop_times_sorted["stop_to_stop_minutes"] > 30
].copy()

print("Number of stop-to-stop gaps above 30 minutes:",
      len(long_gaps))

display(
    long_gaps[
        ["trip_id", "stop_sequence", "arrival",
         "prev_arrival", "stop_to_stop_minutes"]
    ].sort_values(
        "stop_to_stop_minutes",
        ascending=False
    ).head(20)
)
```

```python
# Examine the distribution of stop-to-stop travel times

print("Stop-to-stop travel time distribution:")

print(
    stop_times_sorted["stop_to_stop_minutes"]
    .value_counts()
    .sort_index()
    .tail(20)
)
```

```python
# Investigate long gaps between consecutive stops

long_gaps = stop_times_sorted[
    stop_times_sorted["stop_to_stop_minutes"] > 30
].copy()

# Create a stop_id -> stop_name mapping
stop_name_map = stops.set_index("stop_id")["stop_name"]

# Add stop names without merging
long_gaps["stop_name"] = long_gaps["stop_id"].map(stop_name_map)

# Display the longest gaps
display(
    long_gaps[
        [
            "trip_id",
            "stop_sequence",
            "stop_id",
            "stop_name",
            "prev_arrival",
            "arrival",
            "stop_to_stop_minutes"
        ]
    ]
    .sort_values("stop_to_stop_minutes", ascending=False)
    .head(20)
)
```

```python
# Check whether long-gap stop IDs exist in the stops dataset

long_gap_stop_ids = long_gaps["stop_id"].unique()

missing_stop_ids = [
    stop_id
    for stop_id in long_gap_stop_ids
    if stop_id not in stops["stop_id"].values
]

print("Long-gap stop IDs:", len(long_gap_stop_ids))
print("Missing from stops dataset:", len(missing_stop_ids))
print("Missing stop IDs:", missing_stop_ids)
```

```python
# Check the data types and compare stop IDs

print("stops stop_id type:", stops["stop_id"].dtype)
print("stop_times stop_id type:", stop_times["stop_id"].dtype)

print("\nExample stop IDs from stops:")
print(stops["stop_id"].head(20).tolist())

print("\nExample missing IDs from long gaps:")
print(missing_stop_ids)
```

```python
# Convert stop IDs to the same data type

stops["stop_id"] = stops["stop_id"].astype(str)
stop_times_sorted["stop_id"] = stop_times_sorted["stop_id"].astype(str)

# Recreate the long-gap dataframe
long_gaps = stop_times_sorted[
    stop_times_sorted["stop_to_stop_minutes"] > 30
].copy()

# Add stop names
stop_name_map = stops.set_index("stop_id")["stop_name"]
long_gaps["stop_name"] = long_gaps["stop_id"].map(stop_name_map)

# Display the longest gaps
display(
    long_gaps[
        [
            "trip_id",
            "stop_sequence",
            "stop_id",
            "stop_name",
            "prev_arrival",
            "arrival",
            "stop_to_stop_minutes"
        ]
    ]
    .sort_values("stop_to_stop_minutes", ascending=False)
    .head(20)
)
```

```python
# Count long gaps by stop

long_gap_by_stop = (
    long_gaps
    .groupby(["stop_id", "stop_name"])["stop_to_stop_minutes"]
    .agg(["count", "mean", "max"])
    .sort_values("count", ascending=False)
)

display(long_gap_by_stop.head(20))
```
