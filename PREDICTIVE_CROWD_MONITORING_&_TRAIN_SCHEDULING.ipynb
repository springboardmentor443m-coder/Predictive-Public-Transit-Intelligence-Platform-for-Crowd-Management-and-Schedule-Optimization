# FEATURE ENGINEERING:

Feature engineering is the process of translating raw data into a mathematical language that machine learning algorithms can actually understand. While a human looks at "2023-04-13 18:50:00" and instantly recognizes it as the evening rush hour, an algorithm just sees a meaningless string of text. You have to break that raw data down into explicit, numerical signals.

# Temporal Extraction:

Machine learning models need time broken into distinct categories. You extract the hour, the day of the week, and create binary flags (1 or 0) for business logic, such as whether a specific trip falls inside the morning or evening peak rush.

```python
import numpy as np

# 1. Temporal Extraction
df['Entry_Hour'] = df['Exact_Entry_Timestamp'].dt.hour
df['Day_of_Week'] = df['Exact_Entry_Timestamp'].dt.day_name()
df['Is_Weekend'] = df['Exact_Entry_Timestamp'].dt.weekday.isin([5, 6]).astype(int)
```

```python
# 2. Create the Peak Hour Flag that marks rush hour times with a 1 and all other times with a 0.

df['Is_Peak_Hour'] = df['Entry_Hour'].apply(
    lambda h: 1 if (8 <= h <= 11) or (17 <= h <= 20) else 0
)
```

It flags two peak windows based on the Entry_Hour column:Morning peak: 8:00 AM – 11:59 AM (hours 8–11)Evening peak: 5:00 PM – 8:59 PM (hours 17–20)

# Cyclical Encoding :

Algorithms do not inherently know that Hour 23 (11 PM) and Hour 0 (Midnight) are right next to each other. If you just feed them the numbers 0 to 23, they think midnight and 11 PM are as far apart as possible. You apply " Sine and Cosine transformations " to the hour to teach the model the continuous, circular loop of a 24-hour clock.

```python
# 3. Cyclical Encoding (Teaching the model the 24-hour loop)

df['Hour_Sin'] = np.sin(2 * np.pi * df['Entry_Hour'] / 24.0)
df['Hour_Cos'] = np.cos(2 * np.pi * df['Entry_Hour'] / 24.0)
```

# Domain Metric Calculation

You combine existing columns to engineer the actual targets your model needs to predict. For example, subtracting the Scheduled_Departure_Time from the Actual_Departure_Time gives you the exact Computed_Delay_min.

```python
# 4. Domain Metric Calculation
# Calculate real delay in minutes

df['Computed_Delay_min'] = (
    df['Actual_Departure_Time'] - df['Scheduled_Departure_Time']
).dt.total_seconds() / 60.0
```

# LABEL ENCODING

Label Encoding means converting text into numbers so a computer can understand it. Each unique text category gets its own number.


# EXAMPLE :

Computers only look at math: Machine learning models cannot read words like "Small" or "Large."Keeps the order: It works perfectly for shirts because the numbers match the size order (0 < 1 < 2).

Algorithms cannot multiply or split text like "Yellow Line" or "Rajiv Chowk." Later, right before model training, these strings are converted into binary arrays (One-Hot Encoding) or numerical IDs (Label Encoding).

# You are applying transit logic to combine two raw timestamps into a new, highly relevant numerical feature.

```python
# Calculate platform dwell time (how long the train sits at the station)
df['Dwell_Time_min'] = (
    df['Actual_Departure_Time'] - df['Actual_Arrival_Time']
).dt.total_seconds() / 60.0
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the visual style for the charts
sns.set_theme(style="whitegrid")

# BOXPLOT : Train Occupancy Spikes

plt.figure(figsize=(14, 6))
sns.boxplot(
    data=df,
    x='Entry_Hour',
    y='Train_Occupancy_Count',
    hue='Entry_Hour',       # Added hue to fix the Seaborn warning
    palette="Blues_d",
    legend=False            # Hides the redundant legend
)

plt.title("Distribution: Train Occupancy by Hour", fontsize=16, fontweight='bold')
plt.xlabel("Hour of Day (24-Hour Format)", fontsize=12)
plt.ylabel("Train Occupancy (Passenger Count)", fontsize=12)
plt.tight_layout()
plt.show()
```

```python

# BAR PLOT : Platform Crowd Density Bottlenecks

plt.figure(figsize=(14, 6))
sns.barplot(
    data=df,
    x='Entry_Hour',
    y='Platform_Crowd_Density',
    hue='Entry_Hour',
    palette="mako",
    legend=False,
    errorbar=None  # Removes the error bars for a cleaner look
)

plt.title("Rush Hour Bottlenecks: Average Platform Crowd Density", fontsize=16, fontweight='bold')
plt.xlabel("Hour of Day (24-Hour Format)", fontsize=12)
plt.ylabel("Platform Density (Passengers Waiting)", fontsize=12)

# Add a visual reference line for the "Critical Danger" threshold
plt.axhline(y=800, color='red', linestyle='--', linewidth=2, label="Critical Overcrowding Threshold")
plt.legend(loc="upper left")

plt.tight_layout()
plt.show()
```
